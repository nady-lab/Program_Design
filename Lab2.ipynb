## Q1) Find the distance between two numbers.
Asks the user for two positive integers, storing the integers in variables named a and b.  Fromt these two numbers calculate the distance between them - e.g. if a is 3 and b is 5 the answer should be 2, or if a is 5 and b is 3 the asnwer should aslo be 2.

```python
a = int(input("Give me an integer: "))
b = int(input("Give me another integer: "))

if a > b:
    print(a - b)
else:
    print(b - a)
```

## Q2) Check if a number is positive, negative, or zero.

Write a Python program that asks the user to input an integer (positive or negative whole number) and prints whether a number is positive, negative, or zero.


```python
num = int(input("Give me an integer: "))

if num < 0:
    print("It's negative")
elif num > 0:
    print("It's positive")
else:
    print("It's zero")
```

# Q3) Print the numbers from 1 to n
Write a Python that program that asks the user to input a positive integer (a positive whole number) and store the number in a variable, n. Use a while loop to print out the numbers from 1 to n.

```python
n = int(input("Give me a positive integer: "))
num = 0

while num < n:
    num += 1
    print(num)
```

## Q4) Sum positive integers up to
Write a Python program that asks the user to input a positive integer. Using a while loop, calculate the sum of integers from 1 to the entered integer.  e.g. if the user enters the number 6 the result should be 21 (1+2+3+4+5+6).

```python
n = int(input("Give me a positive integer: "))
total = 0

while n > 0:
    total = total + n
    n = n - 1
print(total)
```

## Q5) Even or odd?
Write a Python program that asks the user to input a positive integer. The program should print "It's even" if the number is even and "It's odd" if it is odd. Can you do it using a only a while loop, if statments and basic arithmetic operations?



```python
n = int(input("Give me a positive integer: "))

if n%2 == 0:
    print("It's even")
else:
    print("It's odd")
```

```python
n = int(input("Give me a positive integer: "))

while n > 1:
    n = n - 2

if n == 1:
    print("It's odd")
else:
    print("It's even")
```
