## Range
Write a Python program that reads in a number and checks if it is between 1 and 100 (inclusive).

```python
num = int(input("Give me a number: "))

if num >= 1 and num <= 100:
    print(""Your number is between 1 and 100."")
else:
    print("Your number is not between 1 and 100.")
```

Can you do it without 'and' or 'or'?

```python
num = int(input("Give me a number: "))

if num in range (1, 101):
    print("Your number is between 1 and 100.")
else:
    print("Your number is not between 1 and 100.")
```
