## Q1
Write a program that reads in a year (positive integer) and returns whether the year is a leap year. A year is a leap year if it is divisible by 4 and not divisible by 100, unless it is divisible by 400, then it is a leap year.

```python
year = int(input("Enter a year: "))

if (year % 4 == 0) or (year % 400 == 0):
    print("It's a leap year!!")
else:
    print("It's not a leap year.")
```

## Q2
Write a program using a loop that asks the user to input a series of positive integers.  Once the user inputs zero the loop should end and the average value of the integers displayed (the average is the sum of all the integers divided by the number of integers the user input before the loop terminates - not including the final zero entered).

```python
x = int(input("Enter a positive int, enter 0 to end program: "))
counter = 0 
sum_x = 0

while x != 0:
    if x > 0:
        counter += 1
        sum_x += x
        x = int(input("Enter a positive int, enter 0 to end program: "))
average = sum_x / counter
print(average)
```

## Q3
Develop an algorithm that uses a loop to compute the sum of the first _n_ (_n_ being a positive integer read in from the user, it must be 1 or greater) terms of the sequence S = 1-3+5-7+9.... Each iteration of the loop should add a term of the sequence. The number _n_ is a positive integer input by the user. The algorithm should explicitly calculate each term of the sequence e.g. if _n_ is 3 the algorithm should add the numbers 1, -3 and 5 using a loop.

```python
n = int(input("Enter the number: "))

if n < 1:
    print("Error")
else:
    total = 0
    term = 1

    for i in range (n):
        if i % 2 == 0:
            total += term
        else:
            total -= term

        term += 2

print("The sum of the first", n,"integers is", total)
```

# Q4
Write a program that asks the user for a number and prints the sum of all even numbers from 1 up to that number (inclusive).

Example: If the user enters 10, the output should be 30.

```python
n = int(input("Enter the number: "))

if n < 1:
    print("Error")
else:
    total = 0
    term = 1

    for i in range(n):
        if term % 2 == 0:
            total += term
            
        term += 1

    print("The sum of the first", n,"even numbers is", total)
```

# Q5
Write a program that asks the user to enter positive integers until they enter -1.
Then print out:

The largest number entered

The smallest number entered

Ignore the -1 in your calculations.

```python
n = int(input("Enter the number: "))
highest = 0
lowest = 0

while n != -1:
    if highest <= n:
        highest = n

    if lowest >= n:
        lowest = n 

    n = int(input("Enter the number: "))

print(highest)
print(lowest)
```

# Q6

Write a program that reads a string from the user and prints the number of vowels in it (a, e, i, o, u).

Example: Input → "Programming" → Output → 3

```python
word = input("Enter the word and i'll tell you how many vowels there are: ")
vowels = "aeiouAEIOU"
count = 0

for char in word:
    if char in vowels:
        count += 1

print(count)
```

# Q7
Write a program that asks the user to enter a number n and then prints a countdown from n to 1, followed by "Blast off!".

```python
n = int(input("Enter num: "))
count = n 

while count > 0:
    for i in range (n):
        print(count)
        count -= 1

print("Blast Off!!")
```

# Q8
Write a program that calculates the factorial of a number entered by the user.

Example: Input → 5 → Output → 120
(Remember: 5! = 5×4×3×2×1)

```python
x = int(input("Enter the number you want to know the factorial of: "))
factorial = x

for i in range (1, x):
    x -= 1
    factorial *= x

print(factorial)
```

# Q10

Write a program that keeps asking the user to guess a secret number until they get it right.
The secret number should be any number you choose (e.g. 7).
When the guess is too low, print "Too low!"
When it’s too high, print "Too high!"
When it’s correct, print "You got it!" and end the loop.

```python
lucky_num = 7
guess = int(input("Guess the lucky number: "))

while guess != lucky_num:
    if guess > lucky_num:
        print("Too high")
    else:
        print("Too low")

    guess = int(input("Guess the lucky number: "))

print("You got it!!!!")
```

# Q11

Ask the user for a positive integer n, then calculate the sum of the squares of all numbers from 1 to n.

Example: n = 3 → 1² + 2² + 3² = 14

```python
n = int(input("Guess the lucky number: "))
n += 1
s_num = 0 
squares = 0

for i in range (1, n):
    squares = i * i
    s_num += squares 

print(s_num)

```
