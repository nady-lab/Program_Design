## Range
Write a Python program that reads in a number and checks if it is between 1 and 100 (inclusive).

```python
number = int(input("Give me an integer: "))

if number <= 100 and number >= 1:
    print("Your number is between 1 and 100")
else:
    print("Your number is not in the range")
```

Can you do it without 'and' or 'or'?

```python
number = int(input("Give me an integer: "))

if number in range(1, 101):
    print("Your number is between 1 and 100")
else:
    print("Your number is not in the range")
```

Can you do it using 'or' without 'and'?

```python
number = int(input("Give me an integer: "))

if number < 1 or number > 100:
    print("Your number is not in the range")
else:
    print("Your number is between 1 and 100")
```

## Print the first n odd numbers using a while loop.
Write a Python program to print the first n (a positive integer read in from the user) odd numbers using a while loop. 

```python
n = int(input("Give me a positive integer: "))

count = 0
i = 1

while count < n:
        print(i)
        i += 2
        count += 1
```

## Print numbers from n down to 1 using a while loop.
Write a Python program to print numbers from some integer, n, down to 1 using a while loop.

```python
n = int(input("Give me a positive integer: "))

counter = n + 1

while counter > 1:
    counter -= 1
    print(counter)
```

## Grade a student based on their marks
Write code that, upon entering a mark (type float), prints a students grade with the following specifications.

- A - 85% or over
- B - 70% or over
- C - 60% or over
- D - 40% or over
- F - less than 40%

Can you do this using only one greater than (>) or less than (<) comparison per grade?


```python
grade = float(input("Give me your grade out of 100: "))

if grade >= 85:
    print("Your grade is: A")
elif grade >= 70:
    print("Your grade is: B")
elif grade >= 60:
    print("Your grade is: C")
elif grade >= 40:
    print("Your grade is: D")
else:
    print("Your grade is: F")
```

# Multiplication Test
Write a program that tests the users multiplication skills. Ask the user to multiply two numbers (generated using the random number generator). Ask the user to do this 5 times, telling them if they are correct or not after each round. Print their score (number of correct answers) after the 5 rounds are complete.

If you are using your own machine and you installed either miniconda or standard python you may need to install the library numpy to generate the random number. It will already be installed on full anaconda.

Miniconda: from the anaconda prompt (normal terminal on mac) type: `conda install numpy`.  <br>Normal python: from the terminal/command window type: `pip install numpy`

Resart your kernel after installing

```python
import numpy as np  # this line imports the library numpy as np
random_number1 = np.random.randint(1, 11) # this uses the function randint to generate a random number between 1 and 10

# Fill in the rest of the code below

max_round = 5
round_num = 0
score = 0

while round_num != max_round:
    random_number1 = np.random.randint(1, 11)
    random_number2 = np.random.randint(1, 11)
    print(random_number1)
    print(random_number2)
    guess = int(input("What's the product of these numbers? "))
    round_num += 1

    if guess == random_number1 * random_number2:
        print("Well Done, Next Round")
        score += 1
    else:
        print("Better Luck Next Time")
        
print("Your score is:",score, "out of 5")
```
