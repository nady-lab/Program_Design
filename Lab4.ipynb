Sort the values in the variables - a having the smallest of the three values, b the middle value and c the largest - using only if statements and assignments.
```python
a = int(input('please enter a value for a\n'))
b = int(input('please enter a value for b\n'))
c = int(input('please enter a value for c\n'))

smallest_value = 0
middle_value = 0
largest_value = 0

if (a <= b and a <= c):
    smallest_value = a
    print(smallest_value)
elif (b <= a and b <= c):
    smallest_value = b
    print(smallest_value)
else:
    smallest_value = c
    print(smallest_value)


if (a >= b and a <= c) or (a >= c and a <= b):
    middle_value = a
    print(middle_value)
elif (b >= a and b <= c) or (b <= a and b >= c):
    middle_value = b
    print(middle_value)
else:
    middle_value = c
    print(middle_value)


if (a >= b and a >= c):
    largest_value = a
    print(largest_value)
elif (b >= a and b >= c):
    largest_value = b
    print(largest_value)
else:
    largest_value = c
    print(largest_value)

print(smallest_value, middle_value, largest_value)
```
